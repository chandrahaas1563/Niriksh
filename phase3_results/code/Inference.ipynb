```python
import os
import numpy as np
import tensorflow as tf

FOLDER_PATH = "/content/drive/MyDrive/Hackathon_phase3_prediction_dataset"
MODEL_PATH  = "/content/mobilenetv3_finetuned.keras"

IMG_SIZE = (128, 128)

class_names = [
    'BRIDGE', 'CLEAN_CRACK', 'CLEAN_LAYER', 'CLEAN_VIA',
    'CMP', 'CRACK', 'LER', 'OPEN', 'OTHERS', 'PARTICLE', 'VIA'
]

SUPPORTED_EXT = ('.png', '.jpg', '.jpeg', '.bmp', '.tif', '.tiff')

print("Loading model …")
model = tf.keras.models.load_model(MODEL_PATH)
print(f"✓ Model loaded: {MODEL_PATH}\n")

def predict_image(path):

    img = tf.image.decode_image(tf.io.read_file(path), channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE)
    arr = np.expand_dims(img.numpy(), axis=0)
    preds = np.squeeze(model.predict(arr, verbose=0))
    return class_names[int(np.argmax(preds))]

image_files = sorted([f for f in os.listdir(FOLDER_PATH) if f.lower().endswith(SUPPORTED_EXT)])

if not image_files:
    print(f"No images found in: {FOLDER_PATH}")
else:
    print(f"Found {len(image_files)} image(s)\n")

    summary = {}
    for fname in image_files:
        try:
            label = predict_image(os.path.join(FOLDER_PATH, fname))
            print(f"{fname}, {label}")
            summary[label] = summary.get(label, 0) + 1
        except Exception as e:
            print(f"{fname}, ERROR: {str(e)}")

    print("\nSUMMARY")
    print("─" * 30)
    for cls, cnt in sorted(summary.items(), key=lambda x: -x[1]):
        print(f"{cls:<20}  {cnt:>5}  ({cnt/len(image_files):.1%})")
    print("─" * 30)
    print(f"{'TOTAL':<20}  {len(image_files):>5}")
    print("Inference complete!")
```
